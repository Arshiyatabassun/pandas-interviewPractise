```python
#Exploratory Data Analysis (EDA)-->used to plot the different visualization  graphs
# 3 steps
# 1:Univaraint Analysis-->Analysing the single variable
# 2:Bivariant Analysis--->Analysing two Variables
# 3.Multi Variant Analysis--->Analysing multiple variables

# To do the analysis we need some library
# Seaborn-->it is one of the most popular visualization library used to plot different graphs

# Different graphs:
# 1.Bar Chart 2.pie chart 3.histogram 4.scatteredplot 5.box plot 6.count plot 7.violin 8.strip 9.lineplot 10 swarmplot 11.pair plot
```

```python
#import the basic libraries

import numpy as np
import pandas as pd
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
import matplotlib.pyplot as plt
```

```python
#seaborn has an builtin dataset
sns.get_dataset_names()
```

```python
#load the iris data  setor read the data

data =sns.load_dataset("iris")
```

```python
data
```

```python
#relational plots:it is used to understand the releationship between two variables
# 1.Line plot 2.scattere plot 3.relplot
```

```python
#line plot:
sns.lineplot(data=data,x='sepal_length',y='sepal_width',color='yellow')
```

```python
#scatter plot:
sns.scatterplot(data=data, x='petal_width',y='sepal_length' ,hue='species' ,palette="rainbow")
```

```python
#relplot:
sns.relplot(data=data,x='petal_width',y='sepal_length' ,kind='line' ,ci=None)
```

```python
#categorical plots
# it is used to analyse one numerical n one categorical column
# 1.count plot 2.bar plot 3.box plot 4.violin plot 5.strip plot 6.swarm plot 7.cat plot
# 1.count plot:basically gives the count of categories
# equal number of records -->balanced data
# unequal number of records -->unbalanced data
sns.countplot(data=data,x='species')
```

```python
# Box plot:
# used to identify the outliers
sns.boxplot(data=data,x='species',y='sepal_length')
```

```python
# Distribution plots:-->2 types-->1.hist plot(only fo numerical)
# 2.dist plot
# used to check the distribution data
```

```python
#1.hist plot(only fo numerical)
sns.histplot(data=data,x='petal_width',kde=True,hue='species')
```

```python
#dist plot=
sns.distplot(x=data['sepal_length'])
```

```python
# pair plot:try to find the releationship bewteen all the variables
sns.pairplot(data=data ,hue='species')
```
